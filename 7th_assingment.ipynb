```python
#1.Smart Light Controller
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"
    def turn_on(self):
        self.status = "ON"
    def turn_off(self):
        self.status = "OFF"
    def display(self):
        print(f"{self.name} is {self.status}")
light_name = input("Light Name: ")
action = input("Action (ON/OFF): ")
light = SmartLight(light_name)
if action.upper() == "ON":
    light.turn_on()
elif action.upper() == "OFF":
    light.turn_off()
else:
    print("Invalid Action!")
light.display()
```

```python
#2.Employee ID Card System
class Employee:
    def __init__(self, name, Eid, department):
        self.name = name
        self.Eid = Eid
        self.department = department
    def display(self):
        print("\nEmployee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.Eid}")
        print(f"Department: {self.department}")
name = input("Employee Name: ")
Eid = input("Employee ID: ")
department = input("Department: ")
employee = Employee(name, Eid, department)
employee.display()
```

```python
#3.Mobile Contact Record
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    def display(self):
        print("\nContact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")
name = input("Contact Name: ")
phone = input("Phone Number: ")
contact = Contact(name, phone)
contact.display()
```

```python
#4. Product Price Tag Generator
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def display(self):
        print(f"\nProduct: {self.name}")
        print(f"Price: ₹{self.price}")
name = input("Product Name: ")
price = float(input("Product Price: ").replace(",", ""))
product = Product(name, price)
product.display()
```

```python
#5.Movie Rating Display System
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    def display(self):
        print(f"\nMovie: {self.name}")
        print(f"Rating: {self.rating:.1f} / 5")
name = input("Movie Name: ")
rating = float(input("Rating (out of 5): "))
movie = Movie(name, rating)
movie.display()
```

```python
#6.Delivery Address Manager
class Delivery:
    def __init__(self, name, address):
        self.name = name
        self.address = address
    def display(self):
        print("\nDelivery Details")
        print(f"Customer: {self.name}")
        print(f"Address: {self.address}")
name = input("Customer Name: ")
address = input("Delivery Address: ")
delivery = Delivery(name, address)
delivery.display()
```
