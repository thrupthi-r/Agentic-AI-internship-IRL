```python
#1.Social Media – Post Engagement Analyzer
def analyze_post_engagement(likes_list):
    total_likes = 0
    for likes in likes_list:
        total_likes += likes
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    print(f"Total Likes: {total_likes}")
    print(f"Post Status: {status}")
likes = [150, 200, 300, 400]  # Total = 1050
analyze_post_engagement(likes)
```

```python
#2.Healthcare – Medicine Stock Alert System
def check_medicine_stock(stock):
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"
    print(f"Medicine Stock: {stock}")
    print(f"Status: {status}")
medicine_stock = 6
check_medicine_stock(medicine_stock)
```

```python
#3.Agriculture – Rainfall Adequacy Checker
def check_rainfall_adequacy(rainfall_data, required_level):
    total_rainfall = 0
    for rainfall in rainfall_data:
        total_rainfall += rainfall
    average_rainfall = total_rainfall / len(rainfall_data)
    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    print(f"Average Rainfall: {int(average_rainfall)}")
    print(f"Rainfall Status: {status}")
seasonal_rainfall = [70, 75, 68, 75] 
required_rainfall = 70
check_rainfall_adequacy(seasonal_rainfall, required_rainfall)
```

```python
#4.Social Media – Duplicate Account Detection
def detect_duplicate_accounts(usernames):
    unique_usernames = set(usernames)
    if len(unique_usernames) != len(usernames):
        result = "Yes"
    else:
        result = "No"
    print(f"Duplicate Accounts Found: {result}")
usernames_list = ["alex", "john", "emma", "alex"]
detect_duplicate_accounts(usernames_list)
```

```python
#5.Healthcare – Appointment Eligibility Checker
def check_appointment_eligibility(age):
    if age >= 18:
        status = "Eligible"
    else:
        status = "Not Eligible"
    print(f"Patient Age: {age}")
    print(f"Eligibility Status: {status}")
patient_age = 21
check_appointment_eligibility(patient_age)
```

```python
#6.Agriculture – Premium Crop Price Filter
def filter_premium_crops(prices, premium_threshold=2000):
    premium_crops = []
    for price in prices:
        if price > premium_threshold:
            premium_crops.append(price)
    print(f"Premium Crops: {premium_crops}")
crop_prices = [1500, 2500, 1800, 3200]
filter_premium_crops(crop_prices)
```

```python
#7.System Monitoring – Application Health Checker
def check_application_health(error_count):
    if error_count == 0:
        status = "Healthy"
    elif error_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    print(f"Error Count: {error_count}")
    print(f"System Status: {status}")
errors = 7
check_application_health(errors)
```

```python
#8.Banking – Daily Transaction Limit Checker
def check_transaction_limit(amount):
    daily_limit = 50000
    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"
    print(f"Transaction Amount: {amount}")
    print(f"Transaction Status: {status}")
transaction_amount = 60000
check_transaction_limit(transaction_amount)
```

```python
#9.E-Learning – Student Attendance Eligibility System
def check_exam_eligibility(attendance_records):
    total_classes = len(attendance_records)
    attended_classes = 0
    for status in attendance_records:
        if status == 1:  
            attended_classes += 1
    attendance_percentage = (attended_classes / total_classes) * 100
    if attendance_percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"
    print(f"Attendance Percentage: {attendance_percentage}")
    print(f"Exam Eligibility: {eligibility}")
attendance = [1, 1, 0, 1, 1]
check_exam_eligibility(attendance)
```

```python
#10.Smart Electricity Bill Analyzer
def calculate_electricity_bill(units):
    bill = 0
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5
    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"
    return bill, usage
units_consumed = 250
total_bill, usage_status = calculate_electricity_bill(units_consumed)
print(f"Units Consumed: {units_consumed}")
print(f"Total Bill: ₹{total_bill}")
print(f"Usage Classification: {usage_status}")
```
