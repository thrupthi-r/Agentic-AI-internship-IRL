```python
#1.Mobile Recharge Validation System
valid_plans= [199, 299, 399, 599]
def recharge_validation(amount):
    if amount < 50:
        print("Minimun Amount is 50")
        return False
    if amount not in valid_plans:
        print("Invalid plan Amount")
        return False
    return True
while True:
    amount=int(input("Enter the Plan Amount "))
    if recharge_validation(amount):
        print("Recharge is successful")
        break
    else:
        print("Try again,Invalid Plan Amount")
        
```

```python
#2.Inventory Reorder Alert System
products={
    "rice":15,
    "Dal":10,
    "sugar":20,
    "oil":25,
    "maida":20
}
def order_check(products):
    for product,quantity in products.items():
        if quantity < 15:
            print(product,"Reorder Alert")
        else:
            print(product,"Stock OK")
order_check(products)
```

```python
#3.Student Result Processing System
def student_result(marks):
    total=0
    for mark in marks:
        total+=mark
        average=total/len(marks)
        if average >50:
            return "Pass"
        else:
            return "Fail"
marks=[45,89,74,63]
result= student_result(marks)
print("Result:",result)
```

```python
#4. Cab Fare Estimator with Retry Option
def cab_fare(distance,peak):
    base_fare=50
    rate_per_km=12
    fare=base_fare+(distance*rate_per_km)
    if peak.lower()=="yes":
        fare=fare*1.25
    return fare
while True:
    distance=float(input("Enter the Distance in km:"))
    peak=input("Is it Peak Hours?,(yes/no):")
    total_fare=cab_fare(distance,peak)
    print("Total Fare :",total_fare)
    again=input("Do you want to calculate again ?,(yes/no):")
    if again.lower()!="yes":
        break
```
