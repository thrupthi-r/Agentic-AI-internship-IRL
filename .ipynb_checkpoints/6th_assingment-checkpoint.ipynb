```python
#1.Smart Parking Lot Management System
def smart_parking(capacity,logs):
    current = 0
    peak = 0
    for log in logs:
        if log.upper() == "IN":
            current += 1
        elif log.upper() == "OUT":
            if current > 0:
                current -= 1
        if current > peak:
            peak = current
    if current >= capacity:
        status = "Full"
    else:
        status = "Available"
    print("\n Parking Summary.")
    print("Currently Parked Vehicles:", current)
    print("Peak Parking Usage:", peak)
    print("Parking Status:", status)
capacity = int(input("Enter Parking Capacity: "))
n = int(input("Enter number of vehicle logs: "))
logs = []
for i in range(n):
    log = input(f"Enter log {i+1} (IN/OUT): ")
    logs.append(log)
smart_parking(capacity,logs)
```

```python
#2.Online Food Delivery Time Estimator
def delivery_time(distance,traffic,weather):
    time = (distance / 30) * 60 
    if traffic.lower() == "low":
        traffic = 5
    elif traffic.lower() == "medium":
        traffic = 10
    elif traffic.lower() == "high":
        traffic = 20
    else:
        traffic = 0
    if weather.lower() == "clear":
        weather = 0
    elif weather.lower() == "rainy":
        weather = 10
    elif weather.lower() == "stormy":
        weather = 20
    else:
        weather = 0
    total= time + traffic + weather
    print("\n Delivery Time Estimate")
    print("Estimated Delivery Time:", round(total), "minutes")
distance = float(input("Enter Distance (km): "))
traffic = input("Enter Traffic Level (Low/Medium/High): ")
weather = input("Enter Weather Condition (Clear/Rainy/Stormy): ")
delivery_time(distance,traffic,weather)
```

```python
#3.Movie Theatre Seat Occupancy Analyzer
def seat_occupancy(total,booked):
    count = len(booked)
    occupancy = (count/total) * 100
    if occupancy == 100:
        status = "Housefull"
        suggestion = "Open additional show"
    elif occupancy >= 75:
        status = "Almost Full"
        suggestion = "Consider opening additional show"
    else:
        status = "Available"
        suggestion = "No additional show needed"
    print("\nShow Analysis.")
    print("Total Seats:", total)
    print("Booked Seats:", count)
    print("Occupancy:", round(occupancy), "%")
    print("Show Status:", status)
    print("Suggestion:", suggestion)
total = int(input("Enter Total Number of Seats: "))
count = int(input("Enter Number of Booked Seats: "))
booked = []
for i in range(count):
    booked.append(1)
seat_occupancy(total,booked)
```

```python
#4.Cloud Server Load Classification System
def server_load(readings):
    total = 0
    for reading in readings:
        total += reading
    average = total / len(readings)
    if average < 50:
        status = "Normal"
    elif 50 <= average < 80:
        status = "Warning"
    else:
        status = "Critical"
    print("\nServer Load Report.")
    print("CPU Readings:", readings)
    print("Average CPU Load:", round(average), "%")
    print("Server Status:", status)
n = int(input("Enter number of CPU readings: "))
readings = []
for i in range(n):
    value = float(input(f"Enter CPU reading {i+1} (%): "))
    readings.append(value)
server_load(readings)
```
